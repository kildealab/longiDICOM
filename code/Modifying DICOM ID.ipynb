```python
import pydicom as dcm
import os
```

```python
def change_id(CT_path, new_ID, output_file_path=''):
    if output_file_path=='':
        output_file_path = CT_path
    if not os.path.exists(output_file_path):
        os.mkdir(output_file_path)

    
    for f in os.listdir(CT_path):
        # Load the DICOM file
        dicom_data = dcm.dcmread(os.path.join(CT_path,f))
        SOP_uid = dicom_data.SOPInstanceUID

        dicom_data.PatientID = new_ID
        dicom_data.PatientName = CT_path.split('/')[-2]
        
        # Save the updated DICOM file
        dicom_data.save_as(os.path.join(output_file_path,'CT.'+SOP_uid+'.dcm'))
```
